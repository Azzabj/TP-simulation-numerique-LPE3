## Partie 1 : Résolution d'EDO

#### Méthode d'Euler : 
La méthode d'Euler utilise une approximation linéaire pour avancer dans le temps. Si nous connaissons la solution $y(t_n)$ à un instant $t_n$, alors la solution à l'instant suivant $t_{n+1} = t_n + h$ est approximée par :
$$
y_{n+1} = y_n + h f(t_n, y_n),
$$
où $h$ est le pas de temps.

#### Méthode de Runge-Kutta d'ordre 4 :
La méthode de Runge-Kutta améliore la précision en évaluant la pente plusieurs fois dans un intervalle avant de calculer $y_{n+1}$. Les étapes sont :
$$
\begin{aligned}
	k_1 & = f(t_n, y_n), \\
	k_2 & = f\left(t_n + \frac{h}{2}, y_n + \frac{h}{2}k_1\right), \\
	k_3 & = f\left(t_n + \frac{h}{2}, y_n + \frac{h}{2}k_2\right), \\
	k_4 & = f(t_n + h, y_n + hk_3).
\end{aligned}
$$
La solution est alors donnée par :
$$
y_{n+1} = y_n + \frac{h}{6}(k_1 + 2k_2 + 2k_3 + k_4).
$$

Les fonctions `euler(f, y0, t0, tf, n)` et `rk4(f, y0, t0, tf, n)` sont déjà implémentées ci-dessous. Elles seront utilisées dans les applications suivantes.

```python
import numpy as np

def euler(f, y0, t0, tf, n):
    h = (tf - t0) / n
    t = np.linspace(t0, tf, n + 1)
    
    # On initialise la liste avec la condition initiale convertie en tableau NumPy
    y = [np.array(y0)]
    
    for i in range(n):
        y_new = y[-1] + h * np.asarray(f(t[i], y[-1]))
        y.append(y_new)
        
    return np.array(y)


def rk4(f, y0, t0, tf, n):
    h = (tf - t0) / n
    t = np.linspace(t0, tf, n + 1)
    
    y = [np.array(y0)]
    
    for i in range(n):
        ti = t[i]
        yi = y[-1]
        
        k1 = np.asarray(f(ti, yi))
        k2 = np.asarray(f(ti + h/2, yi + (h/2) * k1))
        k3 = np.asarray(f(ti + h/2, yi + (h/2) * k2))
        k4 = np.asarray(f(ti + h,   yi + h * k3))
        
        y_new = yi + (h / 6) * (k1 + 2*k2 + 2*k3 + k4)
        y.append(y_new)
        
    return np.array(y)

```

### **Problème 2 : Oscillateur amorti (2ème ordre)**

La dynamique d'un oscillateur amorti est décrite par l'équation différentielle :
$$
m\frac{d^2x}{dt^2} + c\frac{dx}{dt} + kx = 0,
$$
où :
- $x(t)$ : position de l'oscillateur à un instant $t$,
- $m$ : masse de l'oscillateur $kg$,
- $c$ : coefficient d'amortissement $N·s/m$,
- $k$ : constante de raideur $N/m$.

Pour résoudre cette équation, réécrivons-la comme un système d'équations différentielles d'ordre 1 :
$$
\begin{aligned}
\frac{dx}{dt} & = v, \\
\frac{dv}{dt} & = -\frac{k}{m}x - \frac{c}{m}v
\end{aligned}
$$
avec $v = \frac{dx}{dt}$.

##### **Solution analytique : Cas sous-amorti**
Dans ce cas, la solution analytique de l'oscillateur amorti est donnée par :
$$
x(t) = e^{-\alpha t} \left(A \cos(\omega t) + B \sin(\omega t)\right),
$$
où :
$$
\omega = \sqrt{\frac{k}{m} - \left(\frac{c}{2m}\right)^2}.
$$

Les coefficients $ A $ et $ B $ sont déterminés par les conditions initiales :
$$
A = x_0, \quad B = \frac{v_0 + \alpha x_0}{\omega}.
$$

- $ x_0 $ : position initiale $x(0)$,
- $ v_0 $ : vitesse initiale $\frac{dx}{dt}(0)$,
- $ \alpha = \frac{c}{2m} $ : coefficient d'amortissement,
- $ \omega $ : pulsation propre dans le cas sous-amorti.

## Partie 2 : Résolution numérique des équations aux dérivées partielles (EDP)

### 1. Rappel théorique :
Une équation aux dérivées partielles (`EDP`) décrit un phénomène physique dépendant de plusieurs variables indépendantes, notamment l'espace et le temps.

Dans cette partie, nous allons étudier la diffusion thermique dans une barre de longueur `L`. L'évolution de la température est décrite par l'équation de la chaleur en une dimension :

$$
\frac{\partial T}{\partial t} = \alpha \frac{\partial^2 T}{\partial x^2}
$$

où :  
- $T(x, t)$ est la température à l'instant $t$ et à la position $x$.  
- $\alpha$ est le coefficient de diffusion thermique.  
- $x \in [0, L]$, où $L$ est la longueur de la barre étudiée.

### Discrétisation de l'espace et du temps

Pour résoudre numériquement cette équation, on remplace le domaine continu par un ensemble de points régulièrement espacés.

#### Discrétisation spatiale :  
La barre de longueur `L` est divisée en `N` points, séparés par un pas spatial :

$$
\Delta x=\frac{L}{N-1}
$$

Les positions sont données par :
$$
x_i=i\Delta x,\qquad i=0,\ldots,N-1
$$

#### Discrétisation temporelle :  

Le temps est également divisé en intervalles de durée $\Delta t$ :

$$
t_n=n\Delta t
$$

On note alors la température au point `i` et à l'instant `n` :

$$
T_i^n=T(x_i,t_n)
$$ 

### Approximation des dérivées et schéma numérique

En utilisant les différences finies, les dérivées de l'équation de la chaleur sont approximées par :

$$
\frac{\partial T}{\partial t}
\approx
\frac{T_i^{n+1}-T_i^n}{\Delta t}
$$

$$
\frac{\partial^2T}{\partial x^2}
\approx
\frac{T_{i+1}^n-2T_i^n+T_{i-1}^n}{(\Delta x)^2}
$$

En remplaçant ces expressions dans l'équation de la chaleur, on obtient le schéma explicite :

$$
\boxed{
T_i^{n+1}
=
T_i^n+
r\left(T_{i+1}^n-2T_i^n+T_{i-1}^n\right)
}
$$
avec :

$$
\boxed{
r=\frac{\alpha\Delta t}{(\Delta x)^2}
}
$$

Ce schéma permet de calculer la température à chaque point intérieur de la barre à l'instant suivant, à partir des températures à l'instant actuel.

**Condition de stabilité :**

Pour le schéma explicite de l'équation de la chaleur en 1D, la condition de stabilité est :

$$
\boxed{r\leq\frac{1}{2}}
$$

Le choix des pas spatial et temporel doit donc respecter cette condition pour obtenir une simulation stable.
