## 2. Numpy :

Pour utiliser **NumPy**, il faut d'abord importer la bibliothèque :

```python
import numpy as np
```

L'abréviation `np` est une convention couramment utilisée pour désigner `NumPy` dans un programme.

### 2.1 Les tableaux NumPy (array)
L'objet principal de NumPy est le tableau, appelé `ndarray`. Il permet de stocker et de manipuler efficacement des ensembles de nombres.

Un tableau NumPy peut être créé à partir d'une liste Python à l'aide de `np.array()`, et on peut accéder aux éléments du tableau à l'aide de leurs indices, entre crochets. **Remarque :** En Python, l'indexation commence à 0.

```python
a = np.array([2, 3, 4])   # Création d'un tableau 1D
print(a[0])               # Accéder au premier élément
print(a[2])               # Accéder au troisième élément

a[0] = 1                  # Modifier un élément
print(a)                  # Prints "[1, 3, 4]"


b = np.array([[2, 3, 4], [5, 6, 7]])  # Création d'un tableau 2D
print(b[0, 1])                        # Accéder à l'élément situé à la ligne 0, colonne 1

b[1, 1] = 1                           # Modifier un élément
print(b)
print(b[0, 0], b[0, 1], b[1, 0])      # Prints "2 3 5"

c = np.array([(1.5, 2, 3), (4, 5, 6)])# Create a 2D float array
print(c)
print(c[0, 1])                        # Prints "2.0"
```

Les tableaux NumPy possèdent plusieurs **attributs** qui permettent d'obtenir des informations sur leur structure et leurs éléments.

- `ndim` : nombre des axes du tableau.
- `shape` : dimensions du tableau. Pour une matrice de n lignes et m colonnes, shape donne (n,m).
- `size` : nombre total d'éléments du tableau.
- `dtype` : type des éléments du tableau (int, float, etc.).

On peut accéder à ces informations en utilisant la notation `nom_du_tableau.attribut`.

```python
print(a.ndim)             # Prints "1"
print(b.ndim)             # Prints "2"

print(a.shape)            # Prints "(3,)"
print(b.shape)            # Prints "(2, 3)"

print(a.size)             # Prints "3"
print(b.size)             # Prints "6"

print(a.dtype)            # Prints "int64"
print(c.dtype)            # Prints "float64"
```

NumPy propose plusieurs fonctions permettant de créer rapidement des tableaux avec une structure ou des valeurs particulières.

```python
a = np.zeros((2,2))      # crée un tableau rempli de zéros
print(a)                 # Prints "[[ 0.  0.]
                         #          [ 0.  0.]]"

b = np.ones((1,2))       # crée un tableau rempli de uns
print(b)                 # Prints "[[ 1.  1.]]"

c = np.full((2,2), 7)    # crée un tableau rempli d'une valeur donnée
print(c)                 # Prints "[[ 7.  7.]
                         #          [ 7.  7.]]"

d = np.eye(2)            # crée une matrice identité
print(d)                 # Prints "[[ 1.  0.]
                         #          [ 0.  1.]]"
                             
f = np.arange(3)         # créer une suite de nombres de 0 à 2
print(f)                 # Prints "[0 1 2]"

g = np.linspace(2, 10, 5)    # renvoie 5 nombres régulièrement espacés sur l'intervalle 2-10.
print(g)                     # Prints [ 2.  4.  6.  8. 10.]
```

**`np.arange()`** permet de créer une suite de valeurs en précisant le début, la fin et le pas : `np.arange(début, fin, pas)`.      **Remarque:** La valeur finale n'est généralement pas incluse.

**`np.linspace()`** permet de créer un nombre donné de valeurs régulièrement espacées entre deux bornes : `np.linspace(début, fin, nombre_de_valeurs)`.

**Fonctions mathématiques avec Numpy :** Avec NumPy, les fonctions mathématiques telles que sin, cos et exp peuvent être appliquées directement à un tableau. Par exemple, `np.sin()` calcule le sinus de chaque élément d'un tableau.

```python
a = np.arange(3)
print(np.exp(a))       # Exponentielle

print(np.sin(a))       # Sinus

print(np.cos(a))       # Cosinus
```

**Remarque :** `math.sin()` est conçu pour calculer le sinus d'un nombre à la fois, tandis que `np.sin()` peut être appliqué directement à un tableau et calcule le sinus de chacun de ses éléments.

```python
import numpy as np 
import math as m
x = np.array([0, np.pi/2, np.pi]) 
print(np.sin(x))

print(m.sin(0))
print(m.sin(np.pi/2))
print(m.sin(np.pi))
```

On peut modifier l'apparence du graphique à l'aide de différentes options :

- `color` : couleur de la courbe ('blue', 'red', etc.)
